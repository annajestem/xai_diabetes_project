Dataset contains around 768 observations 9 features

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.2,
    stratify = y,
    random_state = 42
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=5000)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    pd.DataFrame(cm),
    annot=True,       # liczby w komórkach
    fmt='d',          # format liczb (integer)
    cmap='Purples'
)

plt.yticks(rotation=0)   # etykiety osi Y pionowo
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score, RocCurveDisplay
import matplotlib.pyplot as plt

roc_auc = roc_auc_score(y_test, model.predict_proba(X_test)[:, 1])
RocCurveDisplay.from_estimator(model, X_test, y_test)
plt.show()
```

Mój wykres ROC jest nie jest idealnie prostokątny, model klasyfikacyjny działa dobrze.
